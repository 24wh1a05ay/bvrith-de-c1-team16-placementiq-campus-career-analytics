```python
%sql

CREATE OR REPLACE TEMP VIEW dq_config AS

SELECT
    CAST('2026-09-01' AS TIMESTAMP)
        AS controlled_snapshot_timestamp,

    'DQ-P16-W06-v1.0'
        AS dq_ruleset_version;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW companies_physical AS

SELECT
    *,
    
    sha2(
        concat_ws(
            '||',
            COALESCE(company_id, ''),
            COALESCE(_run_id, ''),
            COALESCE(_source_file, ''),
            COALESCE(CAST(_candidate_created_at AS STRING), '')
        ),
        256
    ) AS physical_record_key

FROM silver_companies_candidate;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW companies_dq AS

SELECT
    *,
    
    CASE
        WHEN company_id IS NULL
          OR TRIM(company_id) = ''
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-002_check`

FROM companies_physical;
```

```python
%sql

CREATE OR REPLACE TABLE silver_trusted_companies
USING DELTA AS

SELECT
    *,
    'TRUSTED' AS route,
    CAST(ARRAY() AS ARRAY<STRING>) AS failed_rule_ids,
    0 AS failure_count,
    CAST(NULL AS STRING) AS highest_severity,
    current_timestamp() AS dq_checked_at,
    'DQ-P16-W06-v1.0' AS dq_ruleset_version

FROM companies_dq

WHERE `DQ-P16-002_check` = 'PASS';
```

```python
%sql

CREATE OR REPLACE TEMP VIEW skills_physical AS

SELECT
    *,
    
    sha2(
        concat_ws(
            '||',
            COALESCE(skill_id, ''),
            COALESCE(_run_id, ''),
            COALESCE(_source_file, ''),
            COALESCE(CAST(_candidate_created_at AS STRING), '')
        ),
        256
    ) AS physical_record_key

FROM silver_skills_candidate;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW skills_dq AS

SELECT
    *,
    
    CASE
        WHEN skill_id IS NULL
          OR TRIM(skill_id) = ''
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-002_check`,

    CASE
        WHEN proficiency_threshold IS NULL
          OR proficiency_threshold NOT BETWEEN 1 AND 5
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-007_check`

FROM skills_physical;
```

```python
%sql

CREATE OR REPLACE TABLE silver_trusted_skills
USING DELTA AS

SELECT
    *,
    'TRUSTED' AS route,

    CAST(ARRAY() AS ARRAY<STRING>) AS failed_rule_ids,

    0 AS failure_count,

    CAST(NULL AS STRING) AS highest_severity,

    current_timestamp() AS dq_checked_at,

    'DQ-P16-W06-v1.0' AS dq_ruleset_version

FROM skills_dq

WHERE `DQ-P16-002_check` = 'PASS'
  AND `DQ-P16-007_check` = 'PASS';
```

```python
%sql

CREATE OR REPLACE TABLE quarantine_skills
USING DELTA AS

SELECT
    *,
    'QUARANTINE' AS route,

    FILTER(
        ARRAY(
            CASE
                WHEN `DQ-P16-002_check` = 'FAIL'
                THEN 'DQ-P16-002'
            END,

            CASE
                WHEN `DQ-P16-007_check` = 'FAIL'
                THEN 'DQ-P16-007'
            END
        ),
        x -> x IS NOT NULL
    ) AS failed_rule_ids,

    SIZE(
        FILTER(
            ARRAY(
                CASE
                    WHEN `DQ-P16-002_check` = 'FAIL'
                    THEN 'DQ-P16-002'
                END,

                CASE
                    WHEN `DQ-P16-007_check` = 'FAIL'
                    THEN 'DQ-P16-007'
                END
            ),
            x -> x IS NOT NULL
        )
    ) AS failure_count,

    CASE
        WHEN `DQ-P16-002_check` = 'FAIL'
        THEN 'CRITICAL'
        ELSE 'MAJOR'
    END AS highest_severity,

    current_timestamp() AS dq_checked_at,

    'DQ-P16-W06-v1.0' AS dq_ruleset_version

FROM skills_dq

WHERE `DQ-P16-002_check` = 'FAIL'
   OR `DQ-P16-007_check` = 'FAIL';
```

```python
%sql

CREATE OR REPLACE TEMP VIEW students_physical AS

SELECT
    *,
    
    sha2(
        concat_ws(
            '||',
            COALESCE(student_id, ''),
            COALESCE(_run_id, ''),
            COALESCE(_source_file, ''),
            COALESCE(CAST(_candidate_created_at AS STRING), '')
        ),
        256
    ) AS physical_record_key

FROM silver_students_candidate;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW students_dq AS

SELECT
    *,
    
    CASE
        WHEN student_id IS NULL
          OR TRIM(student_id) = ''
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-002_check`

FROM students_physical;
```

```python
%sql

CREATE OR REPLACE TABLE silver_trusted_students
USING DELTA AS

SELECT
    *,
    'TRUSTED' AS route,
    CAST(ARRAY() AS ARRAY<STRING>) AS failed_rule_ids,
    0 AS failure_count,
    CAST(NULL AS STRING) AS highest_severity,
    current_timestamp() AS dq_checked_at,
    'DQ-P16-W06-v1.0' AS dq_ruleset_version

FROM students_dq

WHERE `DQ-P16-002_check` = 'PASS';
```

```python
%sql

CREATE OR REPLACE TEMP VIEW student_skills_physical AS

SELECT

    s.student_id,

    skill.skill_id,

    skill.proficiency_level,

    skill.assessment_score,

    skill.assessed_on,

    s._run_id,

    s._source_file,

    s._ingestion_timestamp,

    sha2(
        concat_ws(
            '||',
            COALESCE(s.student_id, ''),
            COALESCE(skill.skill_id, ''),
            COALESCE(skill.assessed_on, ''),
            COALESCE(CAST(skill.assessment_score AS STRING), ''),
            COALESCE(CAST(skill.proficiency_level AS STRING), ''),
            COALESCE(s._run_id, '')
        ),
        256
    ) AS physical_record_key

FROM `placementiq-campus-career-analytics`.default.silver_students_candidate s

LATERAL VIEW OUTER explode(s.student_skills) skills AS skill;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW student_skills_dq AS

SELECT
    *,

    CASE
        WHEN skill_id IS NULL
          OR TRIM(skill_id) = ''
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-002_check`,

    CASE
        WHEN proficiency_level IS NULL
          OR proficiency_level NOT BETWEEN 1 AND 5
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-007_proficiency_check`,

    CASE
        WHEN assessment_score IS NULL
          OR assessment_score NOT BETWEEN 0 AND 100
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-007_score_check`

FROM student_skills_physical;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW student_skills_routed AS

SELECT
    *,

    FILTER(
        ARRAY(

            CASE
                WHEN `DQ-P16-002_check` = 'FAIL'
                THEN 'DQ-P16-002'
            END,

            CASE
                WHEN `DQ-P16-007_proficiency_check` = 'FAIL'
                  OR `DQ-P16-007_score_check` = 'FAIL'
                THEN 'DQ-P16-007'
            END

        ),
        x -> x IS NOT NULL
    ) AS failed_rule_ids

FROM student_skills_dq;
```

```python
%sql

CREATE OR REPLACE TABLE silver_trusted_student_skills
USING DELTA AS

SELECT
    *,
    'TRUSTED' AS route,
    0 AS failure_count,
    CAST(NULL AS STRING) AS highest_severity,
    current_timestamp() AS dq_checked_at,
    'DQ-P16-W06-v1.0' AS dq_ruleset_version

FROM student_skills_routed

WHERE SIZE(failed_rule_ids) = 0;
```

```python
%sql

CREATE OR REPLACE TABLE quarantine_student_skills
USING DELTA AS

SELECT
    *,
    'QUARANTINE' AS route,

    SIZE(failed_rule_ids) AS failure_count,

    CASE
        WHEN ARRAY_CONTAINS(
            failed_rule_ids,
            'DQ-P16-002'
        )
        THEN 'CRITICAL'
        ELSE 'MAJOR'
    END AS highest_severity,

    current_timestamp() AS dq_checked_at,

    'DQ-P16-W06-v1.0' AS dq_ruleset_version

FROM student_skills_routed

WHERE SIZE(failed_rule_ids) > 0;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW applications_physical AS

SELECT

    s.student_id,

    app.application_id,
    app.application_date,
    app.application_status,
    app.company_id,
    app.final_outcome,
    app.final_outcome_date,
    app.last_updated_timestamp,
    app.rejection_reason,
    app.role_id,
    app.shortlist_flag,

    s._run_id,
    s._source_file,
    s._ingestion_timestamp,

    sha2(
        concat_ws(
            '||',
            COALESCE(s.student_id, ''),
            COALESCE(app.application_id, ''),
            COALESCE(app.application_date, ''),
            COALESCE(app.last_updated_timestamp, ''),
            COALESCE(s._run_id, '')
        ),
        256
    ) AS physical_record_key

FROM silver_students_candidate s

LATERAL VIEW OUTER explode(s.applications) applications AS app;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW applications_dq_001 AS

SELECT
    *,

    CASE
        WHEN application_id IS NULL
          OR TRIM(application_id) = ''
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-001_check`

FROM applications_physical;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW applications_dq_002 AS

SELECT
    *,

    CASE
        WHEN student_id IS NULL
          OR TRIM(student_id) = ''
        THEN 'FAIL'

        WHEN company_id IS NULL
          OR TRIM(company_id) = ''
        THEN 'FAIL'

        WHEN role_id IS NULL
          OR TRIM(role_id) = ''
        THEN 'FAIL'

        ELSE 'PASS'
    END AS `DQ-P16-002_check`

FROM applications_dq_001;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW applications_dq_dates AS

SELECT
    *,

    TRY_CAST(application_date AS TIMESTAMP)
        AS application_ts,

    TRY_CAST(final_outcome_date AS TIMESTAMP)
        AS final_outcome_ts,

    TRY_CAST(last_updated_timestamp AS TIMESTAMP)
        AS last_updated_ts,

    CASE
        WHEN application_date IS NULL
          OR TRY_CAST(application_date AS TIMESTAMP) IS NULL
        THEN 'FAIL'
        ELSE 'PASS'
    END AS application_date_parse_check

FROM applications_dq_002;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW offers_physical AS

SELECT

    s.student_id,

    app.application_id,
    app.application_date,
    app.application_status,
    app.company_id,
    app.final_outcome,
    app.final_outcome_date,
    app.last_updated_timestamp,
    app.role_id,

    offer.offer_id,
    offer.offer_date,
    offer.offer_status,
    offer.offer_type,
    offer.active_flag,
    offer.compensation_band_lpa,

    s._run_id,
    s._source_file,

    sha2(
        concat_ws(
            '||',
            COALESCE(s.student_id, ''),
            COALESCE(app.application_id, ''),
            COALESCE(offer.offer_id, ''),
            COALESCE(offer.offer_date, ''),
            COALESCE(s._run_id, '')
        ),
        256
    ) AS physical_record_key

FROM silver_students_candidate s

LATERAL VIEW OUTER explode(s.applications) applications AS app

LATERAL VIEW OUTER explode(app.offers) offers AS offer;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW final_interviews AS

SELECT
    application_id,

    MAX(
        COALESCE(
            TRY_CAST(actual_end_timestamp AS TIMESTAMP),
            TRY_CAST(scheduled_timestamp AS TIMESTAMP)
        )
    ) AS final_interview_ts

FROM silver_interviews_candidate

WHERE application_id IS NOT NULL

GROUP BY application_id;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW offers_dq_006 AS

SELECT
    o.*,

    fi.final_interview_ts,

    TRY_CAST(o.offer_date AS TIMESTAMP)
        AS offer_ts,

    CASE

        WHEN o.offer_id IS NULL
          OR TRIM(o.offer_id) = ''
        THEN 'FAIL'

        WHEN o.offer_date IS NULL
          OR TRY_CAST(o.offer_date AS TIMESTAMP) IS NULL
        THEN 'FAIL'

        WHEN fi.final_interview_ts IS NOT NULL
         AND TRY_CAST(o.offer_date AS TIMESTAMP)
             < fi.final_interview_ts
        THEN 'FAIL'

        ELSE 'PASS'

    END AS `DQ-P16-006_check`

FROM offers_physical o

LEFT JOIN final_interviews fi

    ON o.application_id = fi.application_id;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW offers_dq_007 AS

SELECT
    *,

    CASE
        WHEN compensation_band_lpa IS NULL
          OR compensation_band_lpa NOT BETWEEN 0 AND 30
        THEN 'FAIL'
        ELSE 'PASS'
    END AS `DQ-P16-007_check`

FROM offers_dq_006;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW application_offer_validity AS

SELECT

    student_id,
    application_id,
    application_status,
    final_outcome,
    final_outcome_date,
    role_id,

    COUNT(
        CASE
            WHEN offer_id IS NOT NULL
             AND TRIM(offer_id) <> ''
             AND UPPER(TRIM(offer_status))
                 IN ('VALID', 'ACCEPTED', 'JOINED')
            THEN 1
        END
    ) AS valid_offer_count

FROM offers_dq_007

GROUP BY
    student_id,
    application_id,
    application_status,
    final_outcome,
    final_outcome_date,
    role_id;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW application_dq_006 AS

SELECT
    *,

    CASE
        WHEN UPPER(TRIM(application_status))
                IN ('ACCEPTED', 'JOINED')
         AND valid_offer_count = 0
        THEN 'FAIL'

        ELSE 'PASS'
    END AS `DQ-P16-006_final_check`

FROM application_offer_validity;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW active_offer_conflicts AS

SELECT

    student_id,
    application_id,

    COUNT(*) AS active_offer_count,

    COLLECT_LIST(
        named_struct(
            'offer_id', offer_id,
            'offer_status', offer_status,
            'offer_date', offer_date,
            'compensation_band_lpa',
                compensation_band_lpa
        )
    ) AS active_offers

FROM offers_dq_007

WHERE UPPER(TRIM(active_flag)) IN ('Y', 'YES', 'TRUE', '1')

GROUP BY
    student_id,
    application_id

HAVING COUNT(*) > 1;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interviews_physical AS

SELECT

    i.*,

    sha2(
        concat_ws(
            '||',
            COALESCE(i.interview_id, ''),
            COALESCE(i.application_id, ''),
            COALESCE(i.student_id, ''),
            COALESCE(i.update_timestamp, ''),
            COALESCE(i._run_id, ''),
            COALESCE(i._source_file, '')
        ),
        256
    ) AS physical_record_key,

    TRY_CAST(i.scheduled_timestamp AS TIMESTAMP)
        AS scheduled_ts,

    TRY_CAST(i.actual_start_timestamp AS TIMESTAMP)
        AS actual_start_ts,

    TRY_CAST(i.actual_end_timestamp AS TIMESTAMP)
        AS actual_end_ts,

    TRY_CAST(i.update_timestamp AS TIMESTAMP)
        AS update_ts

FROM silver_interviews_candidate i;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interviews_with_application AS

SELECT

    i.*,

    TRY_CAST(a.application_date AS TIMESTAMP)
        AS application_ts,

    a.application_status,

    a.final_outcome,

    a.final_outcome_date,

    a.role_id AS application_role_id

FROM interviews_physical i

LEFT JOIN applications_physical a

    ON i.application_id = a.application_id

   AND i.student_id = a.student_id;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_dq_001 AS

SELECT
    *,

    CASE
        WHEN interview_id IS NULL
          OR TRIM(interview_id) = ''
        THEN 'FAIL'

        WHEN physical_record_key IS NULL
        THEN 'FAIL'

        ELSE 'PASS'
    END AS `DQ-P16-001_check`

FROM interviews_with_application;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_dq_002 AS

SELECT
    *,

    CASE

        WHEN student_id IS NULL
          OR TRIM(student_id) = ''
        THEN 'FAIL'

        WHEN company_id IS NULL
          OR TRIM(company_id) = ''
        THEN 'FAIL'

        WHEN role_id IS NULL
          OR TRIM(role_id) = ''
        THEN 'FAIL'

        WHEN application_id IS NULL
          OR TRIM(application_id) = ''
        THEN 'FAIL'

        ELSE 'PASS'

    END AS `DQ-P16-002_check`

FROM interview_dq_001;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_dq_003 AS

SELECT

    i.*,

    CASE

        WHEN scheduled_timestamp IS NULL
          OR scheduled_ts IS NULL
        THEN 'FAIL'

        WHEN application_ts IS NULL
        THEN 'FAIL'

        WHEN scheduled_ts < application_ts
        THEN 'FAIL'

        WHEN scheduled_ts >
             c.controlled_snapshot_timestamp
             + INTERVAL 90 DAYS
        THEN 'FAIL'

        ELSE 'PASS'

    END AS `DQ-P16-003_check`

FROM interview_dq_002 i

CROSS JOIN dq_config c;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_sequence AS

SELECT

    *,

    LAG(round_number) OVER (
        PARTITION BY application_id
        ORDER BY
            scheduled_ts,
            round_number,
            update_ts
    ) AS previous_round_number,

    LAG(scheduled_ts) OVER (
        PARTITION BY application_id
        ORDER BY
            scheduled_ts,
            round_number,
            update_ts
    ) AS previous_scheduled_ts

FROM interview_dq_003;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_dq_004 AS

SELECT

    *,

    CASE

        WHEN round_number IS NULL
        THEN 'FAIL'

        WHEN round_number <= 0
        THEN 'FAIL'

        WHEN previous_round_number IS NOT NULL
         AND round_number < previous_round_number
        THEN 'FAIL'

        WHEN previous_round_number IS NOT NULL
         AND round_number > previous_round_number
         AND scheduled_ts < previous_scheduled_ts
        THEN 'FAIL'

        ELSE 'PASS'

    END AS `DQ-P16-004_check`

FROM interview_sequence;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_dq_005 AS

SELECT

    *,

    CASE

        WHEN UPPER(TRIM(interview_status))
             = 'COMPLETED'

         AND (
              interview_result IS NULL
              OR TRIM(interview_result) = ''
         )

        THEN 'FAIL'


        WHEN UPPER(TRIM(interview_status))
             IN ('CANCELLED', 'NO_SHOW')

         AND UPPER(TRIM(interview_result))
             IN ('SELECTED', 'REJECTED')

        THEN 'FAIL'


        ELSE 'PASS'

    END AS `DQ-P16-005_check`

FROM interview_dq_004;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_versions AS

SELECT

    *,

    ROW_NUMBER() OVER (
        PARTITION BY interview_id
        ORDER BY
            update_ts DESC,
            _ingestion_timestamp DESC
    ) AS version_rank,

    COUNT(*) OVER (
        PARTITION BY interview_id
    ) AS version_count

FROM interview_dq_005;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_conflicts AS

SELECT

    interview_id,

    COUNT(*) AS version_count,

    COUNT(
        DISTINCT CONCAT_WS(
            '||',
            COALESCE(interview_status, ''),
            COALESCE(interview_result, ''),
            COALESCE(company_id, ''),
            COALESCE(role_id, ''),
            COALESCE(CAST(round_number AS STRING), '')
        )
    ) AS distinct_version_values

FROM interviews_physical

WHERE interview_id IS NOT NULL

GROUP BY interview_id

HAVING COUNT(*) > 1;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interview_dq_008 AS

SELECT

    i.*,

    CASE

        WHEN c.interview_id IS NOT NULL
        THEN 'FAIL'

        ELSE 'PASS'

    END AS `DQ-P16-008_check`

FROM interview_dq_005 i

LEFT JOIN interview_conflicts c

    ON i.interview_id = c.interview_id;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interviews_findings AS

SELECT

    *,

    FILTER(
        ARRAY(

            CASE
                WHEN `DQ-P16-001_check` = 'FAIL'
                THEN 'DQ-P16-001'
            END,

            CASE
                WHEN `DQ-P16-002_check` = 'FAIL'
                THEN 'DQ-P16-002'
            END,

            CASE
                WHEN `DQ-P16-003_check` = 'FAIL'
                THEN 'DQ-P16-003'
            END,

            CASE
                WHEN `DQ-P16-004_check` = 'FAIL'
                THEN 'DQ-P16-004'
            END,

            CASE
                WHEN `DQ-P16-005_check` = 'FAIL'
                THEN 'DQ-P16-005'
            END,

            CASE
                WHEN `DQ-P16-008_check` = 'FAIL'
                THEN 'DQ-P16-008'
            END

        ),
        x -> x IS NOT NULL
    ) AS failed_rule_ids

FROM interview_dq_008;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interviews_routed AS

SELECT

    *,

    SIZE(failed_rule_ids)
        AS failure_count,

    CASE

        WHEN ARRAY_CONTAINS(
            failed_rule_ids,
            'DQ-P16-001'
        )
        OR ARRAY_CONTAINS(
            failed_rule_ids,
            'DQ-P16-002'
        )
        OR ARRAY_CONTAINS(
            failed_rule_ids,
            'DQ-P16-008'
        )
        THEN 'CRITICAL'

        WHEN SIZE(failed_rule_ids) > 0
        THEN 'MAJOR'

        ELSE NULL

    END AS highest_severity,

    CASE

        WHEN SIZE(failed_rule_ids) > 0
        THEN 'QUARANTINE'

        ELSE 'TRUSTED'

    END AS route

FROM interviews_findings;
```

```python
%sql

CREATE OR REPLACE TEMP VIEW interviews_physical AS
SELECT
    sha2(
        concat_ws(
            '||',
            COALESCE(interview_id, ''),
            COALESCE(application_id, ''),
            COALESCE(student_id, ''),
            COALESCE(company_id, ''),
            COALESCE(role_id, ''),
            COALESCE(CAST(round_number AS STRING), ''),
            COALESCE(round_name, ''),
            COALESCE(scheduled_timestamp, ''),
            COALESCE(actual_start_timestamp, ''),
            COALESCE(actual_end_timestamp, ''),
            COALESCE(interview_mode, ''),
            COALESCE(interview_status, ''),
            COALESCE(interview_result, ''),
            COALESCE(interviewer_panel_code, ''),
            COALESCE(update_timestamp, ''),
            COALESCE(source_system, ''),
            COALESCE(schema_version, ''),
            COALESCE(_source_file, ''),
            COALESCE(_run_id, '')
        ),
        256
    ) AS physical_record_key,

    *
FROM `placementiq-campus-career-analytics`.default.silver_interviews_candidate;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW interviews_routed AS

WITH base AS (
    SELECT
        *,
        
        TRY_CAST(scheduled_timestamp AS TIMESTAMP) AS scheduled_ts,
        TRY_CAST(actual_start_timestamp AS TIMESTAMP) AS actual_start_ts,
        TRY_CAST(actual_end_timestamp AS TIMESTAMP) AS actual_end_ts,
        TRY_CAST(update_timestamp AS TIMESTAMP) AS update_ts

    FROM interviews_physical
),

checks AS (
    SELECT
        *,

        /* DQ-P16-001
           Critical:
           application_id / interview_id must exist
        */
        CASE
            WHEN interview_id IS NULL
                 OR TRIM(interview_id) = ''
            THEN 'FAIL'

            WHEN application_id IS NULL
                 OR TRIM(application_id) = ''
            THEN 'FAIL'

            ELSE 'PASS'
        END AS dq_p16_001_check,


        /* DQ-P16-003
           Major:
           scheduled timestamp must parse
        */
        CASE
            WHEN scheduled_timestamp IS NULL
                 OR TRIM(scheduled_timestamp) = ''
            THEN 'FAIL'

            WHEN scheduled_ts IS NULL
            THEN 'FAIL'

            ELSE 'PASS'
        END AS dq_p16_003_check,


        /* DQ-P16-004
           Major:
           round_number must be positive
        */
        CASE
            WHEN round_number IS NULL
            THEN 'FAIL'

            WHEN round_number <= 0
            THEN 'FAIL'

            ELSE 'PASS'
        END AS dq_p16_004_check,


        /* DQ-P16-005
           Major:
           invalid status/result combinations
        */
        CASE
            WHEN UPPER(TRIM(interview_status)) = 'COMPLETED'
                 AND (
                     interview_result IS NULL
                     OR TRIM(interview_result) = ''
                 )
            THEN 'FAIL'

            WHEN UPPER(TRIM(interview_status)) IN ('CANCELLED', 'NO_SHOW')
                 AND UPPER(TRIM(interview_result)) IN ('SELECTED', 'REJECTED')
            THEN 'FAIL'

            ELSE 'PASS'
        END AS dq_p16_005_check

    FROM base
),

routed AS (
    SELECT
        *,
        
        ARRAY_REMOVE(
            ARRAY(
                CASE WHEN dq_p16_001_check = 'FAIL' THEN 'DQ-P16-001' END,
                CASE WHEN dq_p16_003_check = 'FAIL' THEN 'DQ-P16-003' END,
                CASE WHEN dq_p16_004_check = 'FAIL' THEN 'DQ-P16-004' END,
                CASE WHEN dq_p16_005_check = 'FAIL' THEN 'DQ-P16-005' END
            ),
            NULL
        ) AS failed_rule_ids,
        
        (
            CASE WHEN dq_p16_001_check = 'FAIL' THEN 1 ELSE 0 END +
            CASE WHEN dq_p16_003_check = 'FAIL' THEN 1 ELSE 0 END +
            CASE WHEN dq_p16_004_check = 'FAIL' THEN 1 ELSE 0 END +
            CASE WHEN dq_p16_005_check = 'FAIL' THEN 1 ELSE 0 END
        ) AS failure_count,
        
        CASE
            WHEN dq_p16_001_check = 'FAIL' THEN 'CRITICAL'
            WHEN (
                CASE WHEN dq_p16_001_check = 'FAIL' THEN 1 ELSE 0 END +
                CASE WHEN dq_p16_003_check = 'FAIL' THEN 1 ELSE 0 END +
                CASE WHEN dq_p16_004_check = 'FAIL' THEN 1 ELSE 0 END +
                CASE WHEN dq_p16_005_check = 'FAIL' THEN 1 ELSE 0 END
            ) > 0 THEN 'MAJOR'
            ELSE NULL
        END AS highest_severity,
        
        CASE
            WHEN (
                CASE WHEN dq_p16_001_check = 'FAIL' THEN 1 ELSE 0 END +
                CASE WHEN dq_p16_003_check = 'FAIL' THEN 1 ELSE 0 END +
                CASE WHEN dq_p16_004_check = 'FAIL' THEN 1 ELSE 0 END +
                CASE WHEN dq_p16_005_check = 'FAIL' THEN 1 ELSE 0 END
            ) > 0 THEN 'QUARANTINE'
            ELSE 'TRUSTED'
        END AS route
        
    FROM checks
)

SELECT * FROM routed;
```

```python
%sql

WITH companies_physical AS (
    SELECT
        sha2(
            concat_ws(
                '||',
                COALESCE(company_id, ''),
                COALESCE(company_alias, ''),
                COALESCE(company_sector, ''),
                COALESCE(company_size_band, ''),
                COALESCE(location_region, ''),
                COALESCE(active_flag, ''),
                COALESCE(role_catalog_json, ''),
                COALESCE(_source_file, ''),
                COALESCE(_run_id, '')
            ),
            256
        ) AS physical_record_key,
        *
    FROM `placementiq-campus-career-analytics`.default.silver_companies_candidate
),
skills_physical AS (
    SELECT
        sha2(
            concat_ws(
                '||',
                COALESCE(skill_id, ''),
                COALESCE(_run_id, ''),
                COALESCE(_source_file, ''),
                COALESCE(CAST(_candidate_created_at AS STRING), '')
            ),
            256
        ) AS physical_record_key,
        *
    FROM `placementiq-campus-career-analytics`.default.silver_skills_candidate
),
student_skills_physical AS (
    SELECT
        sha2(
            concat_ws(
                '||',
                COALESCE(s.student_id, ''),
                COALESCE(skill.skill_id, ''),
                COALESCE(skill.assessed_on, ''),
                COALESCE(CAST(skill.assessment_score AS STRING), ''),
                COALESCE(CAST(skill.proficiency_level AS STRING), ''),
                COALESCE(s._run_id, '')
            ),
            256
        ) AS physical_record_key,
        s.student_id,
        skill.skill_id,
        skill.proficiency_level,
        skill.assessment_score,
        skill.assessed_on,
        s._run_id,
        s._source_file,
        s._ingestion_timestamp
    FROM `placementiq-campus-career-analytics`.default.silver_students_candidate s
    LATERAL VIEW OUTER explode(s.student_skills) skills AS skill
),
reconciliation AS (

    SELECT
        'COMPANIES' AS entity,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM companies_physical
        ) AS candidate_count,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM `placementiq-campus-career-analytics`.default.silver_trusted_companies
        ) AS trusted_count,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM `placementiq-campus-career-analytics`.default.quarantine_companies
        ) AS quarantine_count

    UNION ALL

    SELECT
        'SKILLS' AS entity,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM skills_physical
        ) AS candidate_count,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM `placementiq-campus-career-analytics`.default.silver_trusted_skills
        ) AS trusted_count,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM `placementiq-campus-career-analytics`.default.quarantine_skills
        ) AS quarantine_count

    UNION ALL

    SELECT
        'STUDENT_SKILLS' AS entity,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM student_skills_physical
        ) AS candidate_count,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM `placementiq-campus-career-analytics`.default.silver_trusted_student_skills
        ) AS trusted_count,

        (
            SELECT COUNT(DISTINCT physical_record_key)
            FROM `placementiq-campus-career-analytics`.default.quarantine_student_skills
        ) AS quarantine_count
)

SELECT

    entity,
    candidate_count,
    trusted_count,
    quarantine_count,

    candidate_count
      - trusted_count
      - quarantine_count
        AS variance,

    CASE
        WHEN candidate_count
             =
             trusted_count + quarantine_count
        THEN 'PASS'
        ELSE 'FAIL'
    END AS reconciliation_status

FROM reconciliation

ORDER BY entity;
```
