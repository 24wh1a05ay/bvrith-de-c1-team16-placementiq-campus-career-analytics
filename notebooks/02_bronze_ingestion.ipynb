# Source Inventory

| Source File | Format | Bronze Table |
|-------------|--------|--------------|
| companies.csv | CSV | bronze_companies |
| skills.csv | CSV | bronze_skills |
| students.json | JSON | bronze_students |
| interviews.parquet | Parquet | bronze_interviews |

```python
%sql
LIST '/Volumes/placementiq-campus-career-analytics/default/placement-data/'
```

```python
%sql
CREATE OR REPLACE TEMP VIEW companies_source AS

SELECT *

FROM read_files(
'/Volumes/placementiq-campus-career-analytics/default/placement-data/companies.csv',
format => 'csv',
header => true,
inferSchema => true
);
```

```python
%sql
SELECT *
FROM companies_source
LIMIT 10;
```

```python
%sql
DESCRIBE companies_source
```

```python
%sql
SELECT COUNT(*)
FROM companies_source;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW companies_bronze_ready AS

SELECT
*,

current_timestamp() AS _ingestion_timestamp,

'/Volumes/placementiq-campus-career-analytics/default/placement-data/companies.csv' AS _source_file,

'week4_run01' AS _run_id

FROM companies_source;
```

```python
%sql
CREATE OR REPLACE TABLE bronze_companies

USING DELTA

AS

SELECT *

FROM companies_bronze_ready;
```

```python
%sql
SELECT *
FROM bronze_companies
LIMIT 10;
```

```python
%sql
SELECT COUNT(*)

FROM bronze_companies;
```

```python
%sql
SELECT
'companies' AS file_name,
(SELECT COUNT(*) FROM companies_source) AS source_count,
(SELECT COUNT(*) FROM bronze_companies) AS bronze_count

UNION ALL

SELECT
'skills',
(SELECT COUNT(*) FROM skills_source),
(SELECT COUNT(*) FROM bronze_skills)

UNION ALL

SELECT
'students',
(SELECT COUNT(*) FROM students_source),
(SELECT COUNT(*) FROM bronze_students)

UNION ALL

SELECT
'interviews',
(SELECT COUNT(*) FROM interviews_source),
(SELECT COUNT(*) FROM bronze_interviews);
```

```python
%sql
CREATE OR REPLACE TABLE bronze_companies

USING DELTA

AS

SELECT *

FROM companies_bronze_ready;
```
