```python
%sql
CREATE OR REPLACE TABLE silver_interviews_candidate
USING DELTA
AS
SELECT
  b.*,
  current_timestamp() AS _candidate_created_at,
  'placementiq_silver_candidate_v1.0_baseline' AS _candidate_schema_version
FROM bronze_interviews b;
```

```python
%sql
WITH counts AS (
  SELECT 'interviews' AS entity,
         (SELECT COUNT(*) FROM bronze_interviews) AS bronze_rows,
         (SELECT COUNT(*) FROM silver_interviews_candidate) AS candidate_rows
  UNION ALL
  SELECT 'skills',
         (SELECT COUNT(*) FROM bronze_skills),
         (SELECT COUNT(*) FROM silver_skills_candidate)
  UNION ALL
  SELECT 'students',
         (SELECT COUNT(*) FROM bronze_students),
         (SELECT COUNT(*) FROM silver_students_candidate)
  UNION ALL
  SELECT 'companies',
         (SELECT COUNT(*) FROM bronze_companies),
         (SELECT COUNT(*) FROM silver_companies_candidate)
)
SELECT *,
       candidate_rows - bronze_rows AS difference,
       CASE
         WHEN candidate_rows = bronze_rows THEN 'PASS'
         ELSE 'CHECK'
       END AS status
FROM counts
ORDER BY entity;
```
