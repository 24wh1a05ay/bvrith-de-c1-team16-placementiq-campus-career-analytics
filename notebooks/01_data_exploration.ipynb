# PlacementIQ Week-3B Data Exploration

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()

volume="/Volumes/placementiq-campus-career-analytics/default/placement-data"
display(dbutils.fs.ls(volume))
```

```python
df_students = spark.read.option("multiline", "true").json(f"{volume}/students.json")
df_companies = spark.read.option("header", "true").csv(f"{volume}/companies.csv")
df_skills = spark.read.option("header", "true").csv(f"{volume}/skills.csv")
df_interviews = spark.read.parquet(f"{volume}/interviews.parquet")

display(df_students)
display(df_companies)
display(df_skills)
display(df_interviews)
```

```python
df_students.printSchema()
df_companies.printSchema()
df_skills.printSchema()
df_interviews.printSchema()
```

```python
df_students.createOrReplaceTempView("students")
df_companies.createOrReplaceTempView("companies")
df_skills.createOrReplaceTempView("skills")
df_interviews.createOrReplaceTempView("interviews")
```

```python
spark.sql("""SELECT COUNT(*) AS physical_rows FROM students""").show(truncate=False)
```

```python
spark.sql("""SELECT COUNT(DISTINCT student_id) AS distinct_students FROM students""").show(truncate=False)
```

```python
spark.sql("""SELECT branch,COUNT(*) cnt FROM students GROUP BY branch ORDER BY cnt DESC""").show(truncate=False)
```

```python
spark.sql("""SELECT MIN(graduation_year),MAX(graduation_year) FROM students""").show(truncate=False)
```

```python
spark.sql("""SELECT COUNT(*) missing_student_id FROM students WHERE student_id IS NULL""").show(truncate=False)
```

```python
spark.sql("""SELECT i.company_id FROM interviews i LEFT ANTI JOIN companies c ON i.company_id=c.company_id""").show(truncate=False)
```

```python
spark.sql("""CREATE OR REPLACE TABLE workspace.default.PlacementIQ_week03_bronze_demo_students AS SELECT * FROM students""")
```

```python
spark.sql("""DESCRIBE DETAIL workspace.default.PlacementIQ_week03_bronze_demo_students""").show(truncate=False)
```

```python
spark.sql("""DESCRIBE HISTORY workspace.default.PlacementIQ_week03_bronze_demo_students""").show(truncate=False)
```
